```python
from wpiutil.log import DataLogReader, DataLogEntry, DataLogRecord
from typing import Dict, List


reader = DataLogReader("logs/FRC_20241008_030124.wpilog")

topics: Dict[int, DataLogEntry] = {}

messages: Dict[int, List[DataLogRecord]] = {}

for entry in reader:
    if entry.isStart():
        data = entry.getStartData()
        
        if "gtsam" not in data.name and data.name.startswith("NT:"):
            continue

        topics[data.entry] = data
        messages[data.entry] = []

        continue

    if entry.getEntry() in topics.keys():
        messages[entry.getEntry()].append(entry)

print("Topics:")
for (key, topic) in topics.items():
    print("  " + topic.name)
```
